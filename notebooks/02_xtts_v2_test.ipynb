```python
from TTS.api import TTS

tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2")

print("XTTS v2 loaded successfully")
```

```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/timit_metadata.csv"
)

df.head()
```

```python
speaker_id = "FBCG1"

speaker_data = df[
    df["speaker_id"] == speaker_id
]

speaker_data[
    [
        "speaker_id",
        "utterance_id",
        "transcript",
        "duration",
        "audio_path"
    ]
]
```

```python
reference = speaker_data.iloc[0]

reference_audio = reference["audio_path"]

print("Speaker:", reference["speaker_id"])
print("Reference:", reference_audio)
print("Transcript:", reference["transcript"])
```

```python
text = "This is my first voice cloning experiment."
```

```python
output_path = "../data/generated/xtts_first_test.wav"

tts.tts_to_file(
    text=text,
    speaker_wav=reference_audio,
    language="en",
    file_path=output_path
)

print("Generated:", output_path)
```
