```python
from pathlib import Path
import pandas as pd
import librosa
import soundfile as sf

PROJECT_ROOT = Path.cwd().parent

metadata_path = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "timit_metadata.csv"
)

df = pd.read_csv(metadata_path)

print("Dataset:", df.shape)
```

```python

sample = df.iloc[0]

input_path = PROJECT_ROOT / sample["audio_path"]

audio, sr = librosa.load(
    input_path,
    sr=16000,
    mono=True
)

print("Sample rate:", sr)
print("Duration:", len(audio) / sr)
print("Shape:", audio.shape)
```

```python
import numpy as np

max_value = np.max(np.abs(audio))

if max_value > 0:
    audio = audio / max_value

print("Maximum amplitude:", np.max(np.abs(audio)))
```

```python
output_dir = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "timit_clean"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

output_path = output_dir / "test_sample.wav"

sf.write(
    output_path,
    audio,
    16000
)

print("Saved:", output_path)
```

```python
test_audio, test_sr = librosa.load(
    output_path,
    sr=None,
    mono=True
)

print("Sample rate:", test_sr)
print("Duration:", len(test_audio) / test_sr)
```

```python
processed_files = list(
    clean_output_dir.glob("*.wav")
)

print(
    "Processed audio files:",
    len(processed_files)
)
```

```python
import librosa

successful = 0
failed = 0

for file_path in processed_files[:100]:

    try:
        audio, sr = librosa.load(
            file_path,
            sr=None,
            mono=True
        )

        if sr == 16000 and len(audio) > 0:
            successful += 1
        else:
            failed += 1

    except Exception as e:
        failed += 1
        print("Error:", file_path)
        print(e)

print("Successfully validated:", successful)
print("Failed:", failed)
```
