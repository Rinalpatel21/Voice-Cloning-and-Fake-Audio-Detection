```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/timit_metadata.csv"
)

df.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))

plt.hist(
    df["duration"],
    bins=30
)

plt.xlabel("Duration (seconds)")
plt.ylabel("Number of Audio Files")
plt.title("TIMIT Audio Duration Distribution")

plt.show()
```

```python
dialect_counts = df["dialect"].value_counts().sort_index()

plt.figure(figsize=(10, 6))

dialect_counts.plot(kind="bar")

plt.xlabel("Dialect Region")
plt.ylabel("Number of Recordings")
plt.title("TIMIT Recordings by Dialect")

plt.show()
```

```python
speaker_counts = df["speaker_id"].value_counts()

plt.figure(figsize=(10, 6))

plt.hist(
    speaker_counts,
    bins=10
)

plt.xlabel("Recordings per Speaker")
plt.ylabel("Number of Speakers")
plt.title("TIMIT Recordings per Speaker")

plt.show()
```
